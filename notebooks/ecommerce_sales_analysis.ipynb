### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

# Visualization settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100
```

### Load dataset

```python
df = pd.read_excel("../data/raw/Online Retail.xlsx")
```

```python
df.head()
```

```python
print("Dataset Shape: ", df.shape)
print("-*40")
print("\nColumn Information: ")
df.info()
```

```python
print(df.columns.tolist())
print("\nNull:")
print(df.isnull().sum())
print("\nData types:")
print(df.dtypes)
```

### Missing value percentage

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df)*100
    ).round(2)
})

display(missing_summary)
```

```python
# Duplicate value count
print("Total Duplicate Rows: ", df.duplicated().sum())
```

### Cancelled invoices

```python
cancelled = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
]

print("Cancelled Transactions:", len(cancelled))
print(
    "Cancelled Percentage:",
    round(len(cancelled) / len(df) * 100, 2),
    "%"
)
```

```python
print("Negative Quantity:", (df["Quantity"] < 0).sum())

print("Zero Quantity:", (df["Quantity"] == 0).sum())

print("Negative UnitPrice:", (df["UnitPrice"] < 0).sum())

print("Zero UnitPrice:", (df["UnitPrice"] == 0).sum())
```

### Date range

```python
print("Start Date:", df["InvoiceDate"].min())
print("End Date:", df["InvoiceDate"].max())
```

### Unique values

```python
print("Unique Invoices:", df["InvoiceNo"].nunique())
print("Unique Products:", df["StockCode"].nunique())
print("Unique Customers:", df["CustomerID"].nunique())
print("Unique Countries:", df["Country"].nunique())
```

### Duplicate records inspect

```python
duplicates = df[df.duplicated(keep=False)]

print("Duplicate records including first occurrence:",
      len(duplicates))

display(duplicates.head(10))
```

### Cancelled invoice and negative quantity relationship

```python
cancelled_mask = (
    df["InvoiceNo"].astype(str).str.startswith("C")
)

negative_qty_mask = df["Quantity"] < 0

print("Cancelled rows:", cancelled_mask.sum())

print("Negative quantity rows:", negative_qty_mask.sum())

print("Both cancelled and negative quantity:",
      (cancelled_mask & negative_qty_mask).sum())

print("Negative quantity without C invoice:",
      (~cancelled_mask & negative_qty_mask).sum())
```

### Inspect zero and negative UnitPrice

```python
zero_price = df[df["UnitPrice"] == 0]

negative_price = df[df["UnitPrice"] < 0]

print("Zero price records:", len(zero_price))

display(zero_price.head(10))

print("\nNegative price records:")
display(negative_price)
```

### Inspect negative quantity records

```python
negative_quantity = df[df["Quantity"] < 0]

display(
    negative_quantity[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(15)
)
```

### Cancelled invoice count

```python
print(
    "Unique cancelled invoices:",
    df.loc[cancelled_mask, "InvoiceNo"].nunique()
)
```

### Investigate 1,336 negative quantity records

```python
non_cancelled_returns = df[
    (df["Quantity"] < 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
]

print("Non-cancelled negative quantity:",
      len(non_cancelled_returns))

display(
    non_cancelled_returns[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(20)
)
```

### Zero UnitPrice investigation

```python
zero_price = df[df["UnitPrice"] == 0]

print("Total zero-price records:", len(zero_price))

print("\nMissing descriptions:",
      zero_price["Description"].isnull().sum())

print("\nMost frequent descriptions:")

display(
    zero_price["Description"]
    .value_counts(dropna=False)
    .head(15)
)
```

### Invoice number prefix investigation

```python
df["InvoicePrefix"] = (
    df["InvoiceNo"]
    .astype(str)
    .str.extract(r"^([A-Za-z]+)", expand=False)
    .fillna("Numeric")
)

display(
    df["InvoicePrefix"].value_counts()
)
```

### Duplicate records impact

```python
duplicate_rows = df[df.duplicated()]

print("Duplicate rows:", len(duplicate_rows))

print(
    "Duplicate quantity:",
    duplicate_rows["Quantity"].sum()
)

print(
    "Duplicate revenue:",
    (duplicate_rows["Quantity"] *
     duplicate_rows["UnitPrice"]).sum()
)
```
